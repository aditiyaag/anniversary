```python
import numpy as np
import matplotlib.pyplot as plt

# Inisialisasi data
data = np.array([[5,130],
                 [6, 140],
                 [7, 150],
                 [8, 160],
                 [9, 170],
                 [10, 190]])

# Inisialisasi pusat klaster secara acak
centroid1 = [6, 160]
centroid2 = [8, 180]

# Fungsi untuk menghitung jarak antara dua titik
def euclidean_distance(point1, point2):
    return np.sqrt(np.sum((point1 - point2) ** 2))

# Algoritma K-Means
def k_means(data, centroid1, centroid2, iterations):
    for _ in range(iterations):
        cluster1 = []
        cluster2 = []

        # Assign setiap titik ke klaster terdekat
        for point in data:
            distance_to_centroid1 = euclidean_distance(point, centroid1)
            distance_to_centroid2 = euclidean_distance(point, centroid2)

            if distance_to_centroid1 < distance_to_centroid2:
                cluster1.append(point)
            else:
                cluster2.append(point)

        # Hitung pusat baru dari setiap klaster
        centroid1 = np.mean(cluster1, axis=0)
        centroid2 = np.mean(cluster2, axis=0)

    return cluster1, cluster2, centroid1, centroid2

# Menjalankan algoritma K-Means
cluster1, cluster2, centroid1, centroid2 = k_means(data, centroid1, centroid2, iterations=10)

# Konversi data ke array numpy
cluster1 = np.array(cluster1)
cluster2 = np.array(cluster2)

# Visualisasi grafik
plt.scatter(cluster1[:, 0], cluster1[:, 1], color='blue', label='Cluster 1')
plt.scatter(cluster2[:, 0], cluster2[:, 1], color='red', label='Cluster 2')
plt.scatter(centroid1[0], centroid1[1], color='black', marker='*', s=200, label='Centroid 1')
plt.scatter(centroid2[0], centroid2[1], color='black', marker='*', s=200, label='Centroid 2')
plt.xlabel('Jari-jari (cm)')
plt.ylabel('Tinggi (cm)')
plt.title('K-Means Clustering')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Data
data = np.array([[1.0, 1.3],
                 [1.2, 2.1],
                 [2.3, 2.2],
                 [3.4, 1.2],
                 [1.4, 1.5],
                 [1.2, 2.3],
                 [1.3, 3.0],
                 [1.3, 2.3],
                 [3.5, 3.2],
                 [1.2, 2.1],
                 [1.5, 2.0]])

# Langkah 1: Menentukan Jumlah Klaster
k = 3

# Langkah 2: Membuat Model K-Means
model = KMeans(n_clusters=k, random_state=0)

# Langkah 3: Melatih Model
model.fit(data)

# Langkah 4: Memprediksi Kelas Setiap Titik Data
labels = model.predict(data)

# Langkah 5: Mendapatkan Pusat Klaster
centers = model.cluster_centers_

# Visualisasi Hasil
plt.figure(figsize=(8, 6))

# Memvisualisasikan data pada setiap klaster
for i in range(k):
    cluster_points = data[labels == i]
    plt.scatter(cluster_points[:, 0], cluster_points[:, 1], label=f'Cluster {i+1}')

# Memvisualisasikan pusat klaster
plt.scatter(centers[:, 0], centers[:, 1], color='black', marker='*', s=200, label='Centroids')

plt.xlabel('Fitur 1')
plt.ylabel('Fitur 2')
plt.title('Hasil Klasterisasi K-Means')
plt.legend()
plt.grid(True)
plt.show()
```
